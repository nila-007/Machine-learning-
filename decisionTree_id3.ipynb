```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
data = {
    "CGPA": ["g9", "g8", "g7", "g9", "g6",
             "g8", "g7", "g9", "g6", "g8",
             "g9", "g7", "g8", "g6", "g9", "g7"],

    "INTERACTION": ["Good", "Good", "Poor", "Good", "Poor",
                    "Good", "Poor", "Good", "Poor", "Good",
                    "Good", "Poor", "Good", "Poor", "Good", "Poor"],

    "PK": ["Yes", "Yes", "No", "Yes", "No",
           "Yes", "No", "Yes", "No", "Yes",
           "Yes", "No", "Yes", "No", "Yes", "No"],

    "CS": ["Yes", "No", "No", "Yes", "No",
           "Yes", "No", "Yes", "No", "Yes",
           "Yes", "No", "Yes", "No", "Yes", "No"],

    "Job": ["Yes", "Yes", "No", "Yes", "No",
            "Yes", "No", "Yes", "No", "Yes",
            "Yes", "No", "Yes", "No", "Yes", "No"]
}
table = pd.DataFrame(data)
print("Number of g9 students:")
print((table["CGPA"] == "g9").sum())
encoder=LabelEncoder()
for i in table.columns:
    table[i]=encoder.fit_transform(table[i])
x=table.iloc[:,:4].values
y=table.iloc[:,4].values

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

model = DecisionTreeClassifier(
    criterion="entropy",
   max_depth=3
)

model.fit(x_train, y_train)

y_pred = model.predict(x_test)

print("\nAccuracy:")
print(accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print([1,0,0,1])
if model.predict([[1,0,0,1]])==1:
    print("Got job")
else:
    print("No")
    print([2,0,2,0])
if model.predict([[2,0,2,0]])==1:
    print("Got job")
else:
    print("No")
   
```
