```python
import math
a0 = 5
a1 = 2
a2 = 2
error1 = 2 / math.sqrt(a1**2 + a2**2)
b0 = 3
b1 = 20
b2 = 20
error2 = 2 / math.sqrt(b1**2 + b2**2)
print("Classifier 1:")
print(a0, "+", a1, "x1 +", a2, "x2 = 0")
print("Distance Error =", error1)
print("\nClassifier 2:")
print(b0, "+", b1, "x1 +", b2, "x2 = 0")
print("Distance Error =", error2)
if error1 > error2:
    print("\nClassifier 1 is the good classifier")
elif error2 > error1:
    print("\nClassifier 2 is the good classifier")
else:
    print("\nBoth classifiers are equally good")
```
