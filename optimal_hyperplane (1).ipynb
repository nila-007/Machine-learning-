```python
import matplotlib.pyplot as plt
positive = [(3,1), (3,2), (4,0)]
negative = [(1,0), (0,1), (0,-1)]
for x, y in positive:
    plt.scatter(x, y, marker='o', label='+1' if (x,y) == positive[0] else "")
for x, y in negative:
    plt.scatter(x, y, marker='x', label='-1' if (x,y) == negative[0] else "")
x1 = [0, 3]
x2 = [-2*x + 4.5 for x in x1]
plt.plot(x1, x2, label='Optimal Hyperplane')
plt.scatter(1, 0, s=100, facecolors='none', edgecolors='black')
plt.scatter(3, 1, s=100, facecolors='none', edgecolors='black')
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Optimal Hyperplane")
plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
positive = [(3, 1), (3, 2), (4, 0)]
negative = [(1, 0), (0, 1), (0, -1)]
for x, y in positive:
    plt.scatter(
        x,
        y,
        color="blue",
        marker="o",
        label="+1" if (x, y) == positive[0] else "",
    )
for x, y in negative:
    plt.scatter(
        x,
        y,
        color="red",
        marker="x",
        label="-1" if (x, y) == negative[0] else "",
    )
x1 = np.array([-0.5, 3.5])
x2_optimal = -2 * x1 + 4.5
x2_pos_margin = -2 * x1 + 5.5
x2_neg_margin = -2 * x1 + 3.5
plt.plot(x1, x2_optimal, "k-", label="Optimal Hyperplane")
plt.plot(
    x1,
    x2_pos_margin,
    "b--",
    linewidth=1,
    label="Positive Support Hyperplane (+1)",
)
plt.plot(
    x1,
    x2_neg_margin,
    "r--",
    linewidth=1,
    label="Negative Support Hyperplane (-1)",
)
plt.scatter(
    [3, 1],
    [1, 0],
    s=150,
    facecolors="none",
    edgecolors="black",
    linewidth=1.5,
    label="Support Vectors",
)
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("SVM Optimal Hyperplane & Margin Lines")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.grid(True)
plt.legend(loc="upper right")
plt.xlim(-0.5, 4.5)
plt.ylim(-1.5, 3.0)
plt.show()
```
