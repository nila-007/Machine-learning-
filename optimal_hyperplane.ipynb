```python
import matplotlib.pyplot as plt
positive = [(3,1), (3,2), (4,0)]
negative = [(1,0), (0,1), (0,-1)]
for x, y in positive:
    plt.scatter(x, y, marker='o', label='+1' if (x,y) == positive[0] else "")
for x, y in negative:
    plt.scatter(x, y, marker='x', label='-1' if (x,y) == negative[0] else "")
x1 = [0, 3]
x2 = [-2*x + 4.5 for x in x1]
plt.plot(x1, x2, label='Optimal Hyperplane')
plt.scatter(1, 0, s=100, facecolors='none', edgecolors='black')
plt.scatter(3, 1, s=100, facecolors='none', edgecolors='black')
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Optimal Hyperplane")
plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)
plt.grid(True)
plt.legend()
plt.show()
```
